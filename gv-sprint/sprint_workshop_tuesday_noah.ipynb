## Ziel des Tages

Am Dienstag geht es darum, aus dem bisher definierten Ziel, den Risiken und dem Target erste konkrete Lösungsideen zu entwickeln.

Der Fokus liegt nicht darauf, direkt die finale Lösung zu bauen, sondern verschiedene Ansätze sichtbar zu machen und erste Skizzen für einen möglichen Prototyp zu erstellen.

## 1. Lightning Demos

Ideen und bestehende Lösungen anschauen, die wir übernehmen oder adaptieren können.

### Mögliche Themen für Demos

#### LangGraph
- StateGraph als Grundlage für agentische Workflows
- Nodes und Edges zur Modellierung einzelner Verarbeitungsschritte
- Gemeinsamer State zwischen den Schritten
- Möglichkeit für Rückkopplungsschleifen und Validierung

#### Hub-and-Spoke Architekturen
- Zentraler Orchestrator steuert spezialisierte Worker-Agenten
- Klare Verantwortungstrennung
- Gute Nachvollziehbarkeit und Debuggbarkeit
- Passend für strukturierte mehrstufige Workflows

#### Requirement Engineering Systeme
- Strukturierte Erfassung von Anforderungen
- Eingabemasken oder Templates
- Umwandlung unstrukturierter Nutzerangaben in klar definierte Felder
- Relevante Kategorien wie Kosten, Constraints, Ziele und Parameter

#### Optimierungsmodellierung
- PuLP als mögliche technische Grundlage
- Zielfunktionen
- Nebenbedingungen
- Entscheidungsvariablen
- Beispiele wie Lagerkosten, Mindestbestellmengen oder Servicelevel

## 4. Mögliche Agentenstruktur

### Orchestrator Agent
- empfängt die Nutzereingabe
- steuert den Workflow
- entscheidet, welcher Agent als nächstes aktiv wird
- sammelt die Ergebnisse

### Requirement Agent
- extrahiert Kosten, Constraints, Ziele und Parameter
- strukturiert die Anforderungen
- erkennt fehlende oder widersprüchliche Angaben

### Modeling Agent
- übersetzt strukturierte Anforderungen in ein mathematisches Optimierungsmodell
- definiert Entscheidungsvariablen
- formuliert Zielfunktion und Nebenbedingungen

### Code Agent
- erzeugt ausführbaren Optimierungscode
- zum Beispiel mit PuLP
- bereitet den Code so auf, dass er getestet werden kann

### Validation Agent
- prüft Modell und Code
- erkennt formale Fehler
- gibt Feedback an den Orchestrator
- ermöglicht Rücksprünge im Workflow
